```python
from nltk.corpus import stopwords
stopwords.fileids()
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Abel visited Addis Ababa and works at Google.")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
sent = "I am honored to accept this award on behalf of our team. Their hard work and dedication have been truly remarkable."
print(word_tokenize(sent))
print(sent_tokenize(sent))
```
